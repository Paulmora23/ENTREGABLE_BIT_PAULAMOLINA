# # 1Registro de estudiantes
Crea una lista llamada estudiantes.
Cada elemento de la lista debe ser una tupla con la siguiente estructura:

(nombre, edad, nota)
La lista debe contener mínimo 8 estudiantes.

```python
estudiantes = [
  ("Andrea", 25, 4.8),
  ("Cesar", 19, 3.1 ),
  ("Luis", 23, 3.4),
  ("Daniela", 28, 4.4),
  ("Sofia", 18, 4.5),
  ("Jorge", 28, 2.6),
  ("Claudia", 21, 5.0),
  ("Sebastian", 26, 4.9),
  ("Esmeralda", 17, 2.9),
  ("Natalia", 24, 3.6),
]
```

#  # 2Mostrar la información
Utiliza un ciclo for para recorrer la lista e imprimir la información de cada estudiante con un formato similar al siguiente:

Ana tiene 20 años y obtuvo una nota de 4.5

```python
print("Registro de Estudiantes:")
for estudiante in estudiantes:
  nombre, edad, nota = estudiante
  print(f"{nombre} tiene {edad} años y obtuvo una nota de {nota}")
```

# #3 Clasificación de estudiantes
Utilizando estructuras if, elif y else, clasifica a cada estudiante según su nota.
Nota    Clasificación
Mayor o igual a 4.5    Excelente
Entre 4.0 y 4.49    Bueno
Entre 3.0 y 3.99    Aceptable
Menor a 3.0    Reprobó
Imprime el nombre del estudiante junto con su clasificación.

```python
print("/nClasificación de Estudiantes:")
for nombre,edad,nota in estudiantes:
  if nota >= 4.5:
    clasificacion = "Excelente"
  elif 4.0 <= nota <= 4.49:
    clasificacion = "Bueno"
  elif 3.0 <= nota <= 3.99:
    clasificacion = "Aceptable"
  else:
    clasificacion = "Reprobó"
  print(f"{nombre} obtuvo una nota de {nota} clasificado como {clasificacion}")
```

# # 4Promedio general
Calcula el promedio de todas las notas utilizando un ciclo y una variable acumuladora.
Al finalizar imprime el promedio obtenido.

```python
suma_notas = 0
for nombre, edad, nota in estudiantes:
  suma_notas += nota

promedio= suma_notas / len(estudiantes)
print(f"El promedio de las notas es: {promedio:.2f}")
```

# #5 Búsqueda de estudiante
Solicita al usuario el nombre de un estudiante mediante input().
Si el estudiante existe en la lista, imprime un mensaje indicando que fue encontrado.
Si no existe, informa que no se encontró ningún estudiante con ese nombre.

```python
nombre_buscar = input("Ingrese el nombre del estudiante a buscar: ").strip()

estudiante_encontrado = False
for nombre, edad, nota in estudiantes:
  if nombre.lower() == nombre_buscar.lower():
    estudiante_encontrado = True
    break

if estudiante_encontrado:
    print(f"Estudiante encontrado: {nombre}, {edad} años, nota {nota}")
else:
    print(f"No se encontró ningún estudiante con el nombre '{nombre_buscar}'.")
```

## 8 Ciclo While
Construye un programa que permita ingresar números utilizando un ciclo while.
El programa debe finalizar únicamente cuando el usuario ingrese el número 0.
Al finalizar muestra:
    • Cantidad de números ingresados (sin contar el 0).
    • Suma total.
    • Promedio de los números ingresados.


```python
print("Ingreso de Numeros")
suma_numeros= 0
cantidad_numeros= 0

while True:
  numero= float(input("Ingrese un numero (0 para salir)"))
  if numero == 0:
    break

  suma_numeros += numero
  cantidad_numeros += 1

if cantidad_numeros > 0:
  promedio = suma_numeros / cantidad_numeros
  print(f"Cantidad de numeros ingresados: {cantidad_numeros}")
  print(f"Suma total: {suma_numeros}")
  print(f"Promedio de los numeros ingresados: {promedio:.2f}")
else:
  print("No se ingresaron numeros validos antes del 0")
```

## 9 Uso de break y continue
Recorre los números del 1 al 30.
    • Si el número es múltiplo de 3, utiliza continue.
    • Si el número es igual a 25, utiliza break.
Imprime únicamente los números que fueron procesados por el ciclo.

```python
print ("Numeros del 1 al 30")
numeros_procesados =[]

for i in range(1,31):
  if i == 25:
    break
    if i % 3==0:
      continue
  numeros_procesados.append(i)

print("Numeros procesados:")
print(numeros_procesados)
```
