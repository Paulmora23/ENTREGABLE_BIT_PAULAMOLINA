```python
df[df["Animal name"].isnull()].head(10)
```

Se determina cambiar los valores nulos en la columa de Animal name por "Desconocida".

```python
df["Animal name"]= df["Animal name"].fillna("Desconocida")
```

Se determina eliminar la columna Animal code puesto que no tienen ningun dato del cual nos pueda afectar nuestra base de datos y nos permite simplificar la información.

```python
df = df.drop(columns=['Animal code'])
df.info()
```

#Se determina hayar la mediana de la categoria Weight Kg y Body Lenght cm

```python
mediana_weight = df["Weight kg"].median()
df["Weight kg"] = df["Weight kg"].fillna(mediana_weight)


mediana_length = df["Body Length cm"].median()
df["Body Length cm"] = df["Body Length cm"].fillna(mediana_length)
```

#Corregir valores negativos en medidas numéricas (errores de signos)

```python
df['Weight kg'] = df['Weight kg'].abs()
df['Body Length cm'] = df['Body Length cm'].abs()
```

```python
df['Weight kg'] = df.groupby('Animal type')['Weight kg'].transform(lambda s: s.fillna(s.median()))
df['Body Length cm'] = df.groupby('Animal type')['Body Length cm'].transform(lambda s: s.fillna(s.median()))
```

#Imputar nulos en variables de text

```python
df['Gender'] = df['Gender'].fillna('not determined')
df['Animal name'] = df['Animal name'].fillna('Desconocida')
```

#✅3-Numpy

```python
import numpy as np

# Clasificación biológica según la masa corporal: 'Fauna Mayor' vs 'Fauna Menor'
df['Size_category'] = np.where(df['Weight kg'] >= 5.0, 'Fauna Mayor', 'Fauna Menor')

# Verificación de la distribución de la nueva variable
print("Distribución de la categoría de tamaño:")
display(df['Size_category'].value_counts())
display(df[['Animal type', 'Weight kg', 'Size_category']].head(10))
```

```python
# Cálculo de estadísticos descriptivos con NumPy sobre la variable Weight kg
mediana_peso = np.median(df['Weight kg'])
media_peso = np.mean(df['Weight kg'])
desv_peso = np.std(df['Weight kg'])

print(f"Mediana del peso general: {mediana_peso:.3f} kg")
print(f"Media (promedio) del peso general: {media_peso:.3f} kg")
print(f"Desviación estándar del peso general: {desv_peso:.3f} kg")
```

Con np.where() se creó la variable Size_category, clasificando a los animales en Fauna Mayor ($\ge$ 5 kg; 134 registros) y Fauna Menor (< 5 kg; 691 registros).Al analizar Weight kg con NumPy, la mediana fue de 0.360 kg frente a un promedio de 47.352 kg (desviación estándar: 169.548 kg). Esta diferencia ocurre porque la gran mayoría de la muestra son ardillas y erizos de menos de 1 kg, mientras que el peso extremo de los bisontes distorsiona la media hacia arriba; por ello, la mediana representa mejor el valor central de la base de datos.

#4-✅Análisis con groupby

```python
# Conteo de registros por especie
obs_por_animal = df.groupby('Animal type').size().reset_index(name='Total observaciones')
obs_por_animal = obs_por_animal.sort_values(by='Total observaciones', ascending=False)
display(obs_por_animal)
```

```python
# Promedio de peso (kg) y longitud corporal (cm) por especie
prom_por_animal = df.groupby('Animal type')[['Weight kg', 'Body Length cm']].mean().round(2).reset_index()
display(prom_por_animal)
```

```python
# Promedio de peso y longitud según la ubicación geográfica
prom_por_pais = df.groupby('Country')[['Weight kg', 'Body Length cm']].mean().round(2).reset_index()
prom_por_pais = prom_por_pais.sort_values(by='Weight kg', ascending=False)
display(prom_por_pais)
```

```python
# Análisis adicional: Comparación entre machos y hembras por especie
analisis_genero = df.groupby(['Animal type', 'Gender'])[['Weight kg', 'Body Length cm']].agg(['count', 'mean']).round(2)
display(analisis_genero)
```

Este análisis multidimensional permite identificar dimorfismo sexual evidente en la fauna mayor: los bisontes machos promedian 728.00 kg frente a 481.11 kg de las hembras, mientras que en las ardillas rojas el peso medio entre machos y hembras es prácticamente idéntico (0.30 kg).

**1- Qué se encontró**: La muestra está fuertemente dominada por especies pequeñas.

**2-Qué se encontró:** El peso promedio global no refleja la realidad de la mayoría de los individuos analizados.

**Dimorfismo sexual evidente en grandes mamíferos vs. micromamíferos**

**4-Qué se encontró:** La diferencia de masa corporal entre machos y hembras varía drásticamente según la escala corporal del animal.

- Evidencia: En el bisonte europeo, los machos promedian 728.00 kg frente a los 481.11 kg de las hembras (una brecha de casi 247 kg, ~51% más pesados). Por el contrario, en las ardillas rojas el promedio para ambos sexos es idéntico: 0.30 kg (machos: 0.297 kg vs. hembras: 0.297 kg).
